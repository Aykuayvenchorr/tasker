# Исключение показа в Sidebar автоматически создаваемых ссылок на файлы в папке pages
1. В командной строке создаю новую директорию ***.streamlit***
   ```bash
   mkdir .streamlit
   ```
2. Проверяю ее создание
   ```bash
   ls -lah
   ```
3. Перехожу в нее (если она создана)
   ```bash
   cd .streamlit
   ```
4. Создаю в ней файл ***config.toml***
   ```bash
   nano config.toml
   ```
5. Вписываю в него указание на ***Убрать автоматически создаваемый Sidebar***
   ```toml
   [client]
   showSidebarNavigation = false
   ```
6. Сохраняю `Ctrl+S` и выхожу из файла `Ctrl-X`

# Исключение дублирования таблицы при обновлении страницы после добавления новых данных
Каждый вызов `st.dataframe()` создает **новый элемент** на странице.   
Если код с `st.dataframe()` выполняется заново (например, после нажатия кнопки, без `st.rerun()` или без очистки контейнера), то Streamlit **добавляет новый элемент**, а не заменяет старый.   
Добавление `st.write('')` *разрывает* визуальный блок, и Streamlit перестает воспринимать повторный вызов `st.dataframe()` как добавление в тот же контейнер. Визуально это сбивает старую последовательность рендеринга и **перерисовывает интерфейс заново**.   
Фактически `st.write('')` создает **пустой элемент**, и после него страница выстраивается по-новому.

# Хэш
Для генерации хэша вариант 1:
```python
import hashlib

# Шаг 1: Выбрать алгоритм (например, SHA-256)
hash_obj = hashlib.sha256()

# Шаг 2: Передать данные в байтовом формате
hash_obj.update(b'Nastya')

# Шаг 3: Получить хэш в виде шестнадцатеричной строки
hex_dig = hash_obj.hexdigest()
print(hex_dig)
# Результат: f937268c7e2788ed320ce127ee69a5399592aa33592bdf49fe0780aa886ae633
```
